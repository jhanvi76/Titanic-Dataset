Cleaning, preprocessing, and exploratory data analysis (EDA) on the classic Titanic survival dataset using pandas and seaborn.

```
df["sex"]=df["sex"].replace({"male":1,"female":0})
```

```
df=df.drop("deck",axis=1)
```

```
df= df.fillna(df["age"].mean())
```

```
df["age"]
```

```
df.info()
```

```
df.duplicated().sum()
```

```
df=df.drop_duplicates()
```

```
df.info()
```

```
df["embarked"]
```

```
df=pd.get_dummies(df,columns=["embarked"],drop_first=True)
```

```
df.info()
```

```
df['who'].value_counts()
```

```
df=pd.get_dummies(df,columns=["who"],drop_first=True)
```

```
df.info()
```

```
df.head()
```

```
df['embark_town']
```

```
df['embark_town'].value_counts()
```

```
df['embark_town'].isnull().sum()
```

```
df['embark_town'].nunique()
```

```
print(df['embark_town'].to_string())
```

```
df.info()
```

```
df['embark_town'].value_counts()
```

```
df['embark_town'] = df['embark_town'].fillna(df['embark_town'].mode()[0])
```

```
df['embark_town']
```

```
df['embark_town'].value_counts()
```

```

#copied from gemini to remove some values
mask = df['embark_town'].astype(str).str.contains('29.699')
town_mode = df['embark_town'].mode()[0]

# Replace only those specific rows
df.loc[mask, 'embark_town'] = town_mode
```

```
df['embark_town'].value_counts()
```

```
df=pd.get_dummies(df,columns=['embark_town'],drop_first=True)
```

```
df.info()
```

```
df['class'].value_counts()
```

```
df['class']=df['class'].replace({'First':3,'Second':2,'Third':1})
```

```
df['class']
```

```
df['alive']
```

```
df['alive']=df['alive'].replace({'yes':1,'no':0})
```

```
df['alive']
```

```
df.info()
```

```
df=df.replace({True:1,False:0})
```

```
df.info()
```

```
df.to_csv('Titanic_Cleaned_Final.csv',index=False)
```
